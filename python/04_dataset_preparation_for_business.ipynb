```python
# =============================================================================
# IMPORTACIÓN DE LIBRERÍAS
# =============================================================================

from pathlib import Path

import pandas as pd
```

```python
# =============================================================================
# CONFIGURACIÓN DEL ENTORNO
# =============================================================================

try:
    PROJECT_ROOT = Path.cwd().resolve().parent

    DATA_PATH = PROJECT_ROOT / "data"
    PROCESSED_DATA_PATH = DATA_PATH / "processed"

    print(f"✅ Directorio raíz: {PROJECT_ROOT}")

except Exception as error:
    print(f"❌ Error al configurar el entorno: {error}")
    raise
```

```python
# =============================================================================
# VERIFICACIÓN DE RUTAS
# =============================================================================

print(f"Data: {DATA_PATH}")
print(f"Processed: {PROCESSED_DATA_PATH}")
```

# 4. Carga del dataset enriquecido

## Objetivo

Cargar el dataset enriquecido generado en el Notebook 03 y verificar que la importación se realiza correctamente.

Este dataset constituye el punto de partida para la validación y preparación del modelo analítico definitivo que será utilizado por herramientas de Business Intelligence.

```python
# =============================================================================
# CARGA DEL DATASET ENRIQUECIDO
# =============================================================================

INPUT_PATH = PROCESSED_DATA_PATH / "customer_features.csv"

try:
    customer_dashboard_df = pd.read_csv(INPUT_PATH)

    print("✅ Dataset cargado correctamente.")

except Exception as error:
    print(f"❌ Error al cargar el dataset: {error}")
    raise
```

```python
# =============================================================================
# VERIFICACIÓN DE LA CARGA
# =============================================================================

print(f"Filas: {customer_dashboard_df.shape[0]}")
print(f"Columnas: {customer_dashboard_df.shape[1]}")

display(customer_dashboard_df.head())
```

# 5. Validación del modelo analítico

## Objetivo

```python
# =============================================================================
# VALIDACIÓN DEL MODELO ANALÍTICO
# =============================================================================

print(f"Filas: {customer_dashboard_df.shape[0]}")
print(f"Columnas: {customer_dashboard_df.shape[1]}")

print("\nValores nulos")
display(customer_dashboard_df.isna().sum())

duplicates = customer_dashboard_df.duplicated(
    subset="customer_id"
).sum()

print(f"\nClientes duplicados: {duplicates}")

print("\nTipos de datos")
customer_dashboard_df.info()
```
